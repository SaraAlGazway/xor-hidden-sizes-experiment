# 🧠 XOR Classifier with Different Hidden Sizes (PyTorch)

> This notebook experiments with different hidden layer sizes to solve the XOR problem using PyTorch.

## 🎯 What this notebook covers:

- Defining the XOR dataset
- Testing multiple hidden layer sizes: [1, 2, 4, 8, 16]
- Training a neural network for each configuration
- Comparing accuracy across hidden sizes
- Visualizing the effect of hidden size on learning

## 📊 Expected Results:

- Small hidden sizes (1, 2) → **fail** (~50% accuracy)
- Larger hidden sizes (4, 8, 16) → **succeed** (100% accuracy)
- Best configurations: **4, 8, 16**

```python
#1: Imports
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
#2: Load Data
X = torch.tensor([[0,0], [0,1], [1,0], [1,1]], dtype=torch.float32)
y = torch.tensor([[0], [1], [1], [0]], dtype=torch.float32)
```

```python
#3: TestModel
class TestModel(nn.Module):
    def __init__(self, hidden):
        super().__init__()
        self.layer1 = nn.Linear(2, hidden)
        self.relu = nn.ReLU()
        self.layer2 = nn.Linear(hidden, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.layer1(x)
        x = self.relu(x)
        x = self.layer2(x)
        x = self.sigmoid(x)
        return x
```

### 🏗️ TestModel Architecture

| Layer | Type | Input | Output |
|-------|------|-------|--------|
| 1 | Linear | 2 | hidden |
| 2 | ReLU | hidden | hidden |
| 3 | Linear | hidden | 1 |
| 4 | Sigmoid | 1 | 1 |

**Note:** The `hidden` value changes in each experiment.

```python
#4: Experiment with Different Hidden Sizes
HIDDEN_SIZES = [1, 2, 4, 8, 16]
results = []

for hidden in HIDDEN_SIZES:
    print(f"\n{'='*40}")
    print(f"Testing hidden size: {hidden}")
    print(f"{'='*40}")

    # Build model
    model = TestModel(hidden)
    optimizer = optim.Adam(model.parameters(), lr=0.1)
    criterion = nn.BCELoss()

    # Train
    for epoch in range(500):
        optimizer.zero_grad()
        outputs = model(X)
        loss = criterion(outputs, y)
        loss.backward()
        optimizer.step()

    # Test
    with torch.no_grad():
        preds = model(X).round()
        accuracy = (preds == y).float().mean().item()
        print(f"Hidden={hidden:2d}: Accuracy = {accuracy:.0%}")
        print(f"Predictions: {preds.squeeze().tolist()}")

    results.append((hidden, accuracy, preds.squeeze().tolist()))
```

### 🎓 Conclusion

| Hidden Size | Result |
|-------------|--------|
| 1, 2 | ❌ Failed (~50%) |
| 4, 8, 16 | ✅ Success (100%) |

**Why?**

- XOR is a **non-linear** problem.
- Needs at least **2 neurons** theoretically, **4** practically.
- Increasing hidden size beyond 4 does not improve accuracy further.
